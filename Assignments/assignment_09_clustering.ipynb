## 2. Select and Prepare Features

### Business framing:  

A hotel might want to group guests based on how long they stay, how far in advance they book, or how likely they are to make special requests. You need to pick variables that represent meaningful guest behavior.

### Do the following:
- 2.1 Choose 3–5 numeric features related to customer behavior
- 2.2 Perform basic data preparation - remove outliers, remove duplicates, correct any datatypes that need adjustment
- 2.3 Drop missing values if needed
- 2.4 Standardize using `StandardScaler` for any large values  (over 100)

```python
# Add code here 🔧
# 2.1 Choose numeric features
features = ['lead_time', 'stays_in_weekend_nights', 'stays_in_week_nights', 'adr']

# 2.2 Remove outliers (negative ADR) and duplicates
df_cleaned = df[df['adr'] >= 0].copy()
df_cleaned = df_cleaned.drop_duplicates()

# 2.3 Drop missing values
df_cleaned = df_cleaned.dropna(subset=features)

# 2.4 Standardize features using StandardScaler
scaler = StandardScaler()
scaled_features = scaler.fit_transform(df_cleaned[features])

# Convert to a DataFrame for convenience
df_scaled = pd.DataFrame(scaled_features, columns=features)
display(df_scaled.head())
```

### Reflection:
1. What features did you select and why?
2. What kinds of patterns or segments do you expect to find?

### ✍️ Your Response: 🔧
1. I selected lead_time, stays_in_weekend_nights, stays_in_week_nights, and adr because they directly reflect how customers book and spend. Lead time shows planning behavior, weekend/weeknights capture stay duration patterns, and ADR represents their financial spending profile per night.

2. I expect to find a few clear segments: last-minute weekend travelers who spend more, highly planned weekday business guests, and budget-conscious guests booking far in advance for longer, cheaper stays.


## 3. Apply K-Means Clustering

### Business framing:  

Let’s say you’re working with the hotel’s marketing manager. She wants to group guests into a few clear types to target email campaigns. K-Means is a fast, simple way to try this.

### Do the following:
- Fit a `KMeans` model with your selected features
- Choose a value of `k` (e.g. 3, 4, or 5)
- Predict clusters and assign a cluster number to each guest
- Visualize the clusters using a scatterplot of 2 features
- Do a second visualization using a scatterplot of 2 different features
- Create a summary table to show the averages for each variable, broken out by cluster (this will show which variables stand out for each cluster)

Much of this assignment has already been covered in the lab. Please be sure to complete the lab before the assignment.

```python
# Add code here 🔧
# Fit KMeans
kmeans = KMeans(n_clusters=3, random_state=42)
df_cleaned['kmeans_cluster'] = kmeans.fit_predict(df_scaled)

# Visualize clusters (lead_time vs adr)
plt.figure(figsize=(10, 5))
sns.scatterplot(data=df_cleaned, x='lead_time', y='adr', hue='kmeans_cluster', palette='viridis')
plt.title('K-Means: Lead Time vs ADR')
plt.show()

# Second visualization (weekend vs week nights)
plt.figure(figsize=(10, 5))
sns.scatterplot(data=df_cleaned, x='stays_in_weekend_nights', y='stays_in_week_nights', hue='kmeans_cluster', palette='viridis')
plt.title('K-Means: Weekend vs Week Nights')
plt.show()

# Summary table
summary_kmeans = df_cleaned.groupby('kmeans_cluster')[features].mean()
display(summary_kmeans)
```

### Reflection:
1. What `k` value did you choose, and how did you decide?

2. For each cluster answer the following - what types of customers seem to show up in the cluster?  which variables are most prominent or important for that cluster?  

### ✍️ Your Response: 🔧
1. I chose k=3 because it is simple and captures the main behavioral splits: short-stay shoppers, long-stay guests, and early planners.

2. Cluster 0: Quick planners with low ADR (budget travelers). Cluster 1: High spending, mid-range lead time (leisure travelers). Cluster 2: Long lead time, long stays (planners).


## 4. Apply Gaussian Mixture Model (GMM)

### Business framing:  

Not all guests fit neatly into one cluster. GMM lets us capture uncertainty — useful if customers behave similarly across groups.

### Do the following:
- Fit a GMM with the same number of clusters you chose in Part 3
- Predict soft clusters (remember that soft clustering deals with probabilities, not labels)
- Visualize the clusters using a scatterplot of 2 features
- Do a second visualization using a scatterplot of 2 different features
- Create a summary table to show the averages for each variable, broken out by cluster (this will show which variables stand out for each cluster)

```python
# Add your code here
# Fit GMM
gmm = GaussianMixture(n_components=3, random_state=42)
df_cleaned['gmm_cluster'] = gmm.fit_predict(df_scaled)

# Visualize clusters (lead_time vs adr)
plt.figure(figsize=(10, 5))
sns.scatterplot(data=df_cleaned, x='lead_time', y='adr', hue='gmm_cluster', palette='magma')
plt.title('GMM: Lead Time vs ADR')
plt.show()

# Second visualization (weekend vs week nights)
plt.figure(figsize=(10, 5))
sns.scatterplot(data=df_cleaned, x='stays_in_weekend_nights', y='stays_in_week_nights', hue='gmm_cluster', palette='magma')
plt.title('GMM: Weekend vs Week Nights')
plt.show()

# Summary table
summary_gmm = df_cleaned.groupby('gmm_cluster')[features].mean()
display(summary_gmm)
```

### Reflection:
1. How did the GMM results compare to KMeans?
2. What business questions might GMM help answer better?

### ✍️ Your Response: 🔧
1. GMM produced clusters with different shapes and overlap because it uses probabilities instead of strict circular boundaries, offering softer classification.

2. GMM helps answer which guests are on the borderline between clusters, allowing custom marketing targeting of flexible travelers who fit multiple segments.


## 5. Evaluate Your Models

### Business framing:  

In business, models should be both useful and reliable. You’ll compare model quality using standard evaluation metrics.

### Do the following:
- 5.1 Calculate the following **for each** of the models:
  - WCSS
  - Silhouette Score
  - Davies-Bouldin Index

  **NOTE:** This step may take up to 5 minutes.  It is a lot of computation time.  Please be patient.  Or you can limit the scores to using a random sample of 10K rows.

**Remember**:
- Lower WCSS = tighter, better-defined clusters
- Silhouette score ranges from -1 to 1.  Higher values = better clustering
- Lower Davies-Boulding Index = better clustering

```python
# Add code here 🔧
# Align indices by sampling from df_cleaned first
sample_cleaned = df_cleaned.sample(2000, random_state=42)
sample_indices = sample_cleaned.index

# Standardize only the sample for metric calculation to prevent index misalignment
sampler_scaler = StandardScaler()
sample_scaled = sampler_scaler.fit_transform(sample_cleaned[features])

sample_labels_km = sample_cleaned['kmeans_cluster']
sample_labels_gmm = sample_cleaned['gmm_cluster']

# Metrics
print("KMeans Silhouette:", silhouette_score(sample_scaled, sample_labels_km))
print("KMeans DB Index:", davies_bouldin_score(sample_scaled, sample_labels_km))
print("GMM Silhouette:", silhouette_score(sample_scaled, sample_labels_gmm))
print("GMM DB Index:", davies_bouldin_score(sample_scaled, sample_labels_gmm))
```

### Reflection:

5.3. Would you recommend KMeans or GMM for a business analyst? Why?

### ✍️ Your Response: 🔧
